```python
from datasets import load_dataset

docs = load_dataset(
    "Madras1/rag-qa-fulltext-ptbr",
    "qa_pairs",
    split="train"
)

print(docs)
print(docs[0])
```

```python
df = docs.to_pandas()

doc_questions_df = (
    df.groupby(["document_id", "context"], as_index=False)
      .agg(
          questions=("question", lambda x: "; ".join(x.dropna().astype(str)))
      )
)

doc_questions_df.head()
```

```python
doc_questions_df.to_parquet("data/rag_qa_ptbr_doc_questions.parquet", index=False)
```
